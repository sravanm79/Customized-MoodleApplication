## Task 4: String methods  `[4 marks]`

Write `string_info(s)` that returns a **dictionary** with these keys:
- `"upper"` → the string in uppercase
- `"reversed"` → the string reversed
- `"length"` → number of characters
- `"vowels"` → number of vowels (a, e, i, o, u — count both cases)

````python
def string_info(s):
    return {
        "upper": s.upper(),
        "reversed": s[::-1],
        "length": len(s),
        "vowels": sum(1 for ch in s.lower() if ch in "aeiou"),
    }

# ---------- Do not edit below this line ----------
print(string_info("Python"))
print(string_info("Education"))
````

## Task 5: Initials  `[3 marks]`

Write `initials(full_name)` that returns the capitalised initials, each followed by a dot.
Example: `"ada lovelace"` → `"A.L."`

````python
def initials(full_name):
    return "".join(word[0].upper() + "." for word in full_name.split())

# ---------- Do not edit below this line ----------
print(initials("ada lovelace"))
print(initials("sachin ramesh tendulkar"))
````

## Task 7: Remove duplicates (keep order)  `[3 marks]`

Write `unique_in_order(items)` that returns a new list with duplicates removed,
**keeping the order in which items first appear**.

````python
def unique_in_order(items):
    seen, result = set(), []
    for x in items:
        if x not in seen:
            seen.add(x)
            result.append(x)
    return result

# ---------- Do not edit below this line ----------
print(unique_in_order([3, 1, 3, 2, 1, 4]))
print(unique_in_order(["b", "a", "b", "c"]))
````

**✅ Expected output:**
```text
[3, 1, 2, 4]
['b', 'a', 'c']
```

````python
# ✔️ CHECK — Task 7. Do not edit.
def _test():
    assert unique_in_order([3, 1, 3, 2, 1, 4]) == [3, 1, 2, 4], "order must be preserved"
    assert unique_in_order([]) == [], "empty list should return []"
    assert unique_in_order(["b", "a", "b"]) == ["b", "a"], "should work with strings"

check(7, 3, _test)
````

## Task 8: Dictionaries — word count  `[4 marks]`

Write `word_count(sentence)` that returns a dictionary mapping each word to how many times it appears.
Treat words **case-insensitively** (`"The"` and `"the"` are the same word).

````python
def word_count(sentence):
    counts = {}
    for w in sentence.lower().split():
        counts[w] = counts.get(w, 0) + 1
    return counts

# ---------- Do not edit below this line ----------
print(word_count("The cat and the dog and THE bird"))
````

**✅ Expected output:**
```text
{'the': 3, 'cat': 1, 'and': 2, 'dog': 1, 'bird': 1}
```

````python
# ✔️ CHECK — Task 8. Do not edit.
def _test():
    r = word_count("The cat and the dog and THE bird")
    assert r == {"the": 3, "cat": 1, "and": 2, "dog": 1, "bird": 1}, f"got {r}"
    assert word_count("hi") == {"hi": 1}, "single word case failed"

check(8, 4, _test)
````

## Task 10: Conditionals — grading  `[4 marks]`

Write `grade(marks)` that returns:

| Marks | Grade |
|---|---|
| 90 – 100 | `"A"` |
| 75 – 89 | `"B"` |
| 60 – 74 | `"C"` |
| 40 – 59 | `"D"` |
| below 40 | `"F"` |
| below 0 or above 100 | `"Invalid"` |

````python
def grade(marks):
    if marks < 0 or marks > 100:
        return "Invalid"
    if marks >= 90:
        return "A"
    if marks >= 75:
        return "B"
    if marks >= 60:
        return "C"
    if marks >= 40:
        return "D"
    return "F"

# ---------- Do not edit below this line ----------
for m in [95, 75, 62, 40, 12, 105]:
    print(m, "->", grade(m))
````


## Task 11: For loops — FizzBuzz  `[4 marks]`

Write `fizzbuzz(n)` that **returns a list** for numbers 1 to n where:
- multiples of 3 → `"Fizz"`
- multiples of 5 → `"Buzz"`
- multiples of both → `"FizzBuzz"`
- otherwise → the number itself (as an int)

````python
def fizzbuzz(n):
    result = []
    for i in range(1, n + 1):
        if i % 15 == 0:
            result.append("FizzBuzz")
        elif i % 3 == 0:
            result.append("Fizz")
        elif i % 5 == 0:
            result.append("Buzz")
        else:
            result.append(i)
    return result

# ---------- Do not edit below this line ----------
print(fizzbuzz(15))
````

## Task 12: While loops — sum of digits  `[2 marks]`

Write `sum_digits(n)` that returns the sum of the digits of a non-negative integer
**using a `while` loop** (don't convert to a string).

````python
def sum_digits(n):
    total = 0
    while n > 0:
        total += n % 10
        n //= 10
    return total

# ---------- Do not edit below this line ----------
print(sum_digits(1234))
print(sum_digits(9075))
````

**✅ Expected output:**
```text
10
21
```

````python
# ✔️ CHECK — Task 12. Do not edit.
def _test():
    assert sum_digits(1234) == 10, "sum_digits(1234) should be 10"
    assert sum_digits(0) == 0, "sum_digits(0) should be 0"
    assert sum_digits(99999) == 45, "sum_digits(99999) should be 45"

check(12, 2, _test)
````

## Task 14: Error handling  `[3 marks]`

Write `safe_divide(a, b)` that returns `a / b`, but:
- returns `"Cannot divide by zero"` if `b` is 0
- returns `"Invalid input"` if either value is not a number

Use `try` / `except` — not `if` checks.

````python
def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return "Cannot divide by zero"
    except TypeError:
        return "Invalid input"

# ---------- Do not edit below this line ----------
print(safe_divide(10, 4))
print(safe_divide(10, 0))
print(safe_divide(10, "x"))
````

**✅ Expected output:**


````python
# ✔️ CHECK — Task 14. Do not edit.
def _test():
    assert safe_divide(10, 4) == 2.5, "10 / 4 should be 2.5"
    assert safe_divide(10, 0) == "Cannot divide by zero", "division by zero not handled"
    assert safe_divide("a", 2) == "Invalid input", "non-number input not handled"

check(14, 3, _test)
````

## Task 15: Classes & objects  `[5 marks]`

Create a class `BankAccount` with:
- `__init__(self, owner, balance=0)` storing both values as attributes
- `deposit(amount)` → adds to balance and returns the new balance
- `withdraw(amount)` → subtracts and returns the new balance, but **raises `ValueError("Insufficient funds")`** if amount is more than the balance
- `__str__` → returns e.g. `"Priya: ₹1300"`

````python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        return self.balance

    def withdraw(self, amount):
        if amount > self.balance:
            raise ValueError("Insufficient funds")
        self.balance -= amount
        return self.balance

    def __str__(self):
        return f"{self.owner}: ₹{self.balance}"

# ---------- Do not edit below this line ----------
acc = BankAccount("Priya", 1000)
print(acc.deposit(500))
print(acc.withdraw(200))
try:
    acc.withdraw(5000)
except ValueError as e:
    print("Error:", e)
print(acc)
````

**✅ Expected output:**
```text
1500
1300
Error: Insufficient funds
Priya: ₹1300
```

````python
# ✔️ CHECK — Task 15. Do not edit.
def _test():
    a = BankAccount("Test")
    assert a.owner == "Test" and a.balance == 0, "default balance should be 0"
    assert a.deposit(100) == 100, "deposit should return new balance"
    assert a.withdraw(30) == 70, "withdraw should return new balance"
    try:
        a.withdraw(1000)
        raise AssertionError("withdraw should raise ValueError when funds are insufficient")
    except ValueError:
        pass
    assert str(a) == "Test: ₹70", f"__str__ should give 'Test: ₹70', got {str(a)!r}"

check(15, 5, _test)
````
